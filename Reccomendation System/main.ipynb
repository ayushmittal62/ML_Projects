```python
import pandas as pd
```

```python
data = pd.read_csv("dataset.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
data.columns
```

```python
movies = data[['id', 'title','overview', 'genre']]
```

```python
movies.head(10)
```

```python
movies['tag'] = movies['overview'] + movies['genre']
```

```python
movies.head(5)
```

```python
new_data = movies.drop(columns=['overview', 'genre'])
```

```python
new_data.head(5)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_df= 10000, stop_words='english')
```

```python
vector = cv.fit_transform(new_data['tag'].values.astype('U')).toarray()
```

```python

from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
distance = sorted(list(enumerate(similarity[2])), reverse= True, key=lambda vector:vector[1])
for i in distance[0:5]:
    print(new_data.iloc[i[0]].title)
```

```python
def reccomand(movies):
    index = new_data[new_data['title']==movies].index[0]
    distance = sorted(list(enumerate(similarity[index])), reverse= True, key=lambda vector:vector[1])
    for i in distance[0:5]:
        print(new_data.iloc[i[0]].title)
```

```python
reccomand('Iron Man')
```
